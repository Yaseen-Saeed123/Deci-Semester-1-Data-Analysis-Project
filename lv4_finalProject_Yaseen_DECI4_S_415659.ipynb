# 0. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import random as r
r.seed(67954)
```

# 1. Part 1: Designing A Random Experience

## 1.1. Introduction

* **Experiment** : Rolling The Dice 🎲
* **Process** : We'll simulate the dice rolling process through a python program, changing the number of trials each time

## 1.2. Possible Outcomes & Setting The Target Event

* **Sample Space** (All Possible Outcomes): {1, 2, 3, 4, 5, 6}
* **Target Event** : Check if the outcome number is a **prime number**
    - The numbers that satisfy the target event are : {2, 3, 5}

## 1.3. Calculate The Theoretical Probability

**The Theoretical Probability of target event** = $\frac{\text{No. of favourable outcomes}}{\text{Total sample space}}$ = $\frac{3}{6}$ = $\frac{1}{2}$ = **0.5**

## 1.4. Set The Number Of Trials & Run The Experiment

We will perform this experiment three times, each with a different number of trials:
  - First Experiment : 10 Trials
  - Second Experiment : 30 Trials
  - Third Experiment : 100 Trials

##### **Design A Function To Run The Experiment with changing the number of trials each time**

```python
def dice_rolling(n,file_name):
  sample_space = [1,2,3,4,5,6]
  target = [2,3,5]
  trial_rows = []
  count = 0
  print("Trial No | Outcome | Did it achieve target")
  for i in range(n):
    is_target = "No"
    choice = r.choice(sample_space)
    if choice in target:
      is_target = "Yes"
      count += 1
    print(f"Trial {i+1} | Outcome : {choice} | Target Event : {is_target}")
    trial_rows.append({
      'Outcome': choice,
      'Target Event Result (Yes/No)': is_target
    })
  print("-"*30)
  df = pd.DataFrame(trial_rows)
  print(f"Number of Target Events : {count}")
  print(f"Experimental Probability : {count / n}")
  df.to_csv(file_name, index=False)
  return (count / n)
```

##### **First Experiment : 10 Trials**

```python
print(f"Number of Trials : 10")
print("-"*30)
prob = dice_rolling(10, "Experiment 1.csv")
```

##### **Second Experiment : 30 Trials**

```python
print(f"Number of Trials : 30")
print("-"*30)
prob2 = dice_rolling(30,"Experiment 2.csv")
```

##### **Third Experiment: 100 Trials**

```python
print(f"Number of Trials : 100")
print("-"*30)
prob3 = dice_rolling(100, "Experiment 3.csv")
```

## 1.5. Compare Theoretical Probability and Experimental Probabilities

```python
data_dict = {
    'Experiment': ['Exp 1', 'Exp 1', 'Exp 2', 'Exp 2', 'Exp 3', 'Exp 3'],
    'Type': ['Theoretical', 'Experimental', 'Theoretical', 'Experimental', 'Theoretical', 'Experimental'],
    'Probability': [0.5, prob, 0.5, prob2, 0.5, prob3]
}
df = pd.DataFrame(data_dict)
sns.set_style('whitegrid')
sns.set_palette('viridis')
ax = sns.barplot(x='Experiment', y='Probability', hue='Type', data=df)
for container in ax.containers:
    ax.bar_label(container, fmt='%.2f', padding=3)
plt.title('Theoretical vs. Experimental Probabilities')
plt.xlabel('Experiment')
plt.ylabel('Probability')
plt.ylim(0, 1)
plt.yticks(np.arange(0, 1.1, 0.1))
plt.legend(title='Type')
plt.show()
```

This grouped bar chart shows that the experimental probability gets closer to the theoretical probability as the number of trials increases

## 1.6. Make A Master Dataframe Having All Experiments

```python
master_df = pd.concat([
    pd.read_csv("Experiment 1.csv"),
    pd.read_csv("Experiment 2.csv"),
    pd.read_csv("Experiment 3.csv")
],
ignore_index=True)
```

## 1.7. Divide The Master Dataframe into 4 equal groups of 35 datapoint each

```python
def group_data(i):
  group_df = master_df.iloc[i*35:(i+1)*35]
  return group_df
# Set The Groups
group_1 = group_data(0)
group_2 = group_data(1)
group_3 = group_data(2)
group_4 = group_data(3)
```

## 1.8. Calculate Number of Successful Outcomes & Experimental Probability for each Group

```python
def count(group, no, st, end):
    success = (group['Target Event Result (Yes/No)'] == 'Yes').sum()
    prob = round(success / 35, 2)
    print(f"Group {no}")
    print("."*5)
    print(f"Number Of Successful Outcomes : {success}")
    print(f"Trial Range : {st} ~ {end}")
    print("Total Trials : 35")
    print(f"Experimental Probability : {prob}")
    print("-"*30)
    return prob

# Print Out the summary for each group
g1 = count(group_1, 1, 1, 35)
g2 = count(group_2, 2, 36, 70)
g3 = count(group_3, 3, 71, 105)
g4 = count(group_4, 4, 106, 140)
```

This Multivariant scatterplot proves the following:
> Students who study more hours tend to secure higher marks on both previous and current exam scores

## 2.9. Data Story: How To Improve Students' Exam Scores

Improving Students' Exam Scores is a crucial objective for any educational organization; because it's a key metric for the success of both the organization & the students. But before discussing this topic, we need to understand what truly impacts the exam scores in order to take proper action.

### Understanding the variables in a student's academic year
Most people believe that students' performances on exams differ due to two main features, which are:
1.   Student Lifestyle Variables (Sleep Hours, Screen Time, ..etc)
2.   Student Academic Variables (Homework Completion, Attendance Rate, Study Hours/Day, Previous Exam Scores, Studying Method ..etc)

And it's a common belief that lifestyle affects Exam Scores greatly, much more than academic variables.
<br><br>
But, On analysing the dataset given, it turns out that Lifestyle Variables do not affect the Exam Scores as much as the academic variables do. For example, Screen Time only changes by **1.18 hours** from the usage of a Student who achieved an **Excellent** result to one who **Needs improvement**.The difference is too minor to make an impact on Exam Scores. <br><br>
Similarly, Sleep hours do not have a major impact on Exam Scores. According to Our dataset, It's not necessary that when you sleep more, you get higher marks.
<br><br>
Also, not all academic variables have a major impact on Exam Scores. For Instance, Studying Method doesn't have a significant impact on the final grade as almost all studying methods share the same average exam score (About 90 out of 100) more or less.
<br><br>
Adding to that, Attendance Rate doesn't affect Exam Scores. More Attendance doesn't necessarily contribute to higher grades.

> **The bottom Line: The belief that lifestyle variables have a major impact on student's exam score is incorrect**

### So, What does actually affect Exam Scores?
According to our analysis, we have discovered that both `Previous Scores` & `Study Hours` are the variables that have the most impact on student's exam scores. **Here is how:**<br><br>
During The EDA phase, we discovered that Students who scored well on the previous exam will likely do well on this exam, too. Here is where the `Previous Scores` come into play. Also, it was discovered that Students who study more hours tend to score higher.<br><br>
To visually prove it, we have made this `Multivariate Scatterplot` that puts all these factors into action:<br><br>
image.png

On the opposite corner (which indicates lower marks), we find it packed with blue dots which indicates less study hours. Also, the red correlation line shows a strong positive relationship between `Previous Scores` & `Current Scores`.

### So, What should we do?
To improve students' exam scores we should do the following:
1. **Stop exaggerating the minor effect of lifestyle variables**
2. **Always monitor previous marks and make a quick academic intervention to students who got bad marks**
3. **Persuade students to dedicate more time for studying.**
